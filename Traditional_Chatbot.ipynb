```python
import random
from datetime import datetime
```

```python
print("=" * 60)
print("      Traditional Student Assistant Chatbot")
print("=" * 60)
print("Type 'help' to see my capabilities.")
print("Type 'bye' anytime to exit.")
print()
```

```python
jokes = [
    "Why did the programmer quit his job? Because he didn't get arrays.",
    "Python programmers don't byte—they just code.",
    "Debugging is like being the detective in a mystery where you are also the culprit."
]

capabilities = [
    "Say Hello",
    "Tell my name",
    "Tell today's date",
    "Tell current time",
    "Tell a joke",
    "Explain my capabilities",
    "Tell who developed me",
    "Respond to thanks",
    "Exit the chatbot"
]
```

```python
def chatbot():

    while True:

        user = input("You: ").lower().strip()

        if user in ["hello", "hi", "hey"]:

            print("\nBot: Hello! Welcome to the Student Assistant Chatbot.\n")

        elif "name" in user:

            print("\nBot: My name is Student Assistant Bot.\n")

        elif "developer" in user or "created" in user:

            print("\nBot: I was developed as a traditional rule-based chatbot using Python.\n")

        elif "time" in user:

            print("\nBot:", datetime.now().strftime("%I:%M %p"))
            print()

        elif "date" in user:

            print("\nBot:", datetime.now().strftime("%B %d, %Y"))
            print()

        elif "joke" in user:

            print("\nBot:", random.choice(jokes))
            print()

        elif "thank" in user:

            print("\nBot: You're welcome!\n")

        elif "help" in user or "capabilities" in user:

            print("\nI can perform the following tasks:\n")

            for i, item in enumerate(capabilities, start=1):
                print(f"{i}. {item}")

            print()

        elif user in ["bye", "quit", "exit"]:

            print("\nBot: Goodbye! Have a wonderful day.\n")
            break

        else:

            print("\nBot: Sorry, I don't understand that question.\n")
```

```python
chatbot()
```
