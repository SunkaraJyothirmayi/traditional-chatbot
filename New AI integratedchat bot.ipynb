```python
import requests
import random
from datetime import datetime
```

```python
API_TOKEN = "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"

API_URL = "https://api-inference.huggingface.co/models/google/flan-t5-base"

headers = {
    "Authorization": f"Bearer {API_TOKEN}"
}
```

```python
def ask_ai(question):

    payload = {
        "inputs": question
    }

    response = requests.post(
        API_URL,
        headers=headers,
        json=payload
    )

    if response.status_code == 200:

        result = response.json()

        if isinstance(result, list):
            return result[0]["generated_text"]

        return str(result)

    return f"Connection Error ({response.status_code})"
```

```python
def chatbot():

    jokes = [
        "Why do programmers prefer Python? Because it is easy to read.",
        "Debugging is like being a detective in a mystery where you are also the culprit.",
        "Why did the computer get cold? It forgot to close Windows."
    ]

    print("="*60)
    print("AI Integrated Student Assistant Chatbot")
    print("="*60)
    print("Type 'bye' to exit.\n")

    while True:

        user = input("You: ").lower().strip()

        if user in ["hello", "hi", "hey"]:

            print("Bot: Hello! Welcome.\n")

        elif "name" in user:

            print("Bot: I am an AI Integrated Student Assistant Chatbot.\n")

        elif "time" in user:

            print("Bot:", datetime.now().strftime("%I:%M %p"))
            print()

        elif "date" in user:

            print("Bot:", datetime.now().strftime("%B %d, %Y"))
            print()

        elif "joke" in user:

            print("Bot:", random.choice(jokes))
            print()

        elif "help" in user:

            print("""
I can:

- Say Hello
- Tell my name
- Tell the date
- Tell the time
- Tell jokes
- Answer AI questions
- Exit
""")

        elif user in ["bye","exit","quit"]:

            print("Bot: Goodbye!")
            break

        else:

            print("\nBot (AI):")

            answer = ask_ai(user)

            print(answer)
            print()
```
